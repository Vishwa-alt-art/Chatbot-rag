```python
balance = 5000
retries = 0
if retries == 3:
    print("Card Blocked")
    retries = 0
else:
    enter_pin = input("Enter PIN:")
    if enter_pin == "1234":
        print("\n==WELCOME TO UNION BANK ATM===")

        daily_limit = 10000
        history = []

        while True:

           

            print("\n ===ATM ===")
            print("1.Check Balance")
            print("2. Deposit")
            print("3. Withdraw")
            print("4. Exit")

            choice = input("Enter choice: ")

            if choice == "1":
                print("Balance:",balance)

            elif choice == "2":
                amount = int(input("Enter the amount: "))
                balance += amount
                history.append(f'Deposit {amount}')
                print('Deposited amount:', amount)

            elif choice == "3":
                with_amount = int(input("Enter amount: "))
                if with_amount > daily_limit:
                    print("Daily Limit Exceeded")
                if with_amount <= balance:
                    balance -= with_amount
                    history.append(f'Withdraw {with_amount}')
                    print("Withdrawn:",with_amount)

                else:
                    print("Insufficient Balance")

            elif choice == "4":
                print("Thank You")
                break
            else:
                print("Invalid Choice")
            print("\n===Mini Statement===")
            print(f"Deposit{amount}")
            print(f"Withdraw{with_amount}")
            print(f"Current_Balance{balance}")
else:
    retries += 1
    
```
